# Restart vs Interrupt

### Interrupt Kernel
- Stops only the currently running code.
- Keeps existing variables and memory.
- Useful when a cell is taking too long or has an infinite loop.

### Restart Kernel
- Stops execution and clears all variables and memory.
- Useful when the notebook is in an inconsistent state or when testing reproducibility.
